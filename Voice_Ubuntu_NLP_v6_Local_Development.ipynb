# Voice Ubuntu NLP v6 — Local Development and Testing

Open this notebook from inside the extracted `Voice_Ubuntu_NLP_v6` project folder after installing `requirements-dev.txt`. This notebook inspects predictions and parser output; it never executes Ubuntu actions.

```python
from pathlib import Path
import sys, json, pandas as pd
PROJECT_DIR = Path.cwd()
assert (PROJECT_DIR / "voice_ubuntu_intent_model.joblib").is_file(), "Open this notebook from the project folder."
sys.path.insert(0, str(PROJECT_DIR))
from nlp_runtime import predict_intent
from voice_ubuntu_parser import parse_text
print("Project:", PROJECT_DIR)
```

```python
training = pd.read_csv(PROJECT_DIR / "training_examples.csv")
print("Training examples:", len(training))
print("Intent classes:", training["intent"].nunique())
training["intent"].value_counts().sort_index()
```

```python
commands = [
    "Open Firefox",
    "Locate my project PDF",
    "Check RAM",
    "Create a folder in Documents called Notes",
    "Play Interstellar",
    "Play Interstellar on YouTube",
    "ممكن تفتح يوتيوب",
    "شغل أغنية تملي معاك",
    "ابحث على يوتيوب عن شرح التفاضل",
    "Tell me a joke",
]
for text in commands:
    print("\nINPUT:", text)
    print("CLASSIFIER:", json.dumps(predict_intent(text), ensure_ascii=False))
    print("PARSER:", json.dumps(parse_text(text), ensure_ascii=False, indent=2))
```
